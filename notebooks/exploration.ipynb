# 🎬 Movie Recommendation System — Exploration

This notebook walks through EDA and model experimentation.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from backend.data.sample_data import get_movies_df, get_ratings_df, get_all_genres
```

## 1. Dataset Overview

```python
movies = get_movies_df()
ratings = get_ratings_df()
print(f'Movies: {len(movies)}')
print(f'Ratings: {len(ratings)}')
print(f'Users: {ratings.user_id.nunique()}')
movies.head()
```

```python
ratings.describe()
```

## 2. Rating Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
ratings['rating'].hist(bins=20, ax=axes[0], color='#e8c547', edgecolor='black')
axes[0].set_title('Rating Distribution')
axes[0].set_xlabel('Rating')

ratings.groupby('user_id').size().hist(bins=15, ax=axes[1], color='#f07240', edgecolor='black')
axes[1].set_title('Ratings per User')
axes[1].set_xlabel('Number of Ratings')
plt.tight_layout()
plt.show()
```

## 3. Genre Distribution

```python
from collections import Counter
genre_counts = Counter(g for genres in movies['genres'] for g in genres)
gdf = pd.DataFrame(genre_counts.most_common(), columns=['Genre', 'Count'])
gdf.plot(kind='barh', x='Genre', y='Count', legend=False, color='#e8c547', figsize=(8,6))
plt.title('Movies per Genre')
plt.tight_layout()
plt.show()
```

## 4. Collaborative Filtering

```python
from backend.models.collaborative import CollaborativeFilter
cf = CollaborativeFilter().fit(ratings, movies)
recs = cf.recommend(user_id=1, n=10)
pd.DataFrame(recs)[['title', 'genres', 'predicted_rating']]
```

## 5. Content-Based Filtering

```python
from backend.models.content_based import ContentBasedFilter
cb = ContentBasedFilter().fit(movies)
recs = cb.recommend_by_movie('Inception', n=8)
pd.DataFrame(recs)[['title', 'genres', 'similarity_score']]
```
